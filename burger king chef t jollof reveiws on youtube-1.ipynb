```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

headers = {"User-Agent": "Mozilla/5.0"}

url = "https://www.google.com/search?q=Burger+King+Jollof+Rice+review"

response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")

reviews = []
for item in soup.find_all("span", class_="review-text"):
    reviews.append(item.get_text())

df = pd.DataFrame(reviews, columns=["Review"])
df.to_csv("google_jollof_reviews.csv", index=False)
print(df.head())
```

```python
import pandas as pd
from textblob import TextBlob

# Load your scraped comments
df = pd.read_csv("burger_king_jollof_youtube_comments.csv")

# Remove empty comments
df = df[df["Comment"].str.strip() != ""]
df = df.dropna(subset=["Comment"])

# Sentiment function
def get_sentiment(text):
    score = TextBlob(str(text)).sentiment.polarity
    if score > 0.1:    return "Positive"
    elif score < -0.1: return "Negative"
    else:              return "Neutral"

def get_score(text):
    return round(TextBlob(str(text)).sentiment.polarity, 2)

# Apply sentiment
df["Sentiment"] = df["Comment"].apply(get_sentiment)
df["Score"]     = df["Comment"].apply(get_score)

# Results
print("=== SENTIMENT SUMMARY ===")
print(df["Sentiment"].value_counts())
print(f"\nTotal Comments:    {len(df)}")
print(f"Positive Reviews:  {len(df[df['Sentiment']=='Positive'])}")
print(f"Negative Reviews:  {len(df[df['Sentiment']=='Negative'])}")
print(f"Neutral Reviews:   {len(df[df['Sentiment']=='Neutral'])}")
print(f"Overall Score:     {df['Score'].mean():.2f} (-1 worst, +1 best)")

# Save final analysed file
df.to_csv("burger_king_jollof_analysed.csv", index=False)
print("\nAnalysed file saved ✅")
```

```python
import pandas as pd
from textblob import TextBlob
from langdetect import detect, LangDetectException
```

```python
def detect_language(text):
    try:
        lang = detect(str(text))
        return lang
    except LangDetectException:
        return "unknown"

def pidgin_sentiment(text):
    text_lower = str(text).lower()
    
    neg_count = sum(1 for kw in pidgin_negative if kw in text_lower)
    pos_count = sum(1 for kw in pidgin_positive if kw in text_lower)
    neu_count = sum(1 for kw in pidgin_neutral  if kw in text_lower)

    if neg_count > pos_count:
        return "Negative", round(-0.5 - (neg_count * 0.1), 2)
    elif pos_count > neg_count:
        return "Positive", round(0.5 + (pos_count * 0.1), 2)
    elif neu_count > 0:
        return "Neutral", 0.0
    else:
        return "Neutral", 0.0

def english_sentiment(text):
    score = TextBlob(str(text)).sentiment.polarity
    if score > 0.1:    return "Positive", round(score, 2)
    elif score < -0.1: return "Negative", round(score, 2)
    else:              return "Neutral",   round(score, 2)

def analyse_comment(text):
    if not str(text).strip():
        return "Neutral", 0.0, "unknown"

    lang = detect_language(text)

    # Nigerian Pidgin often detected as Afrikaans (af)
    # or Indonesian (id) or unknown by langdetect
    if lang in ["en"]:
        sentiment, score = english_sentiment(text)
    else:
        # Check for Pidgin keywords first
        text_lower = str(text).lower()
        has_pidgin = any(
            kw in text_lower
            for kw in pidgin_positive + pidgin_negative + pidgin_neutral
        )
        if has_pidgin:
            sentiment, score = pidgin_sentiment(text)
            lang = "pidgin"
        else:
            # Fall back to TextBlob for other languages
            sentiment, score = english_sentiment(text)

    return sentiment, score, lang

print("✅ Sentiment analyser ready!")
```

```python
import shutil
import os

desktop = os.path.expanduser("~/Desktop")

project_files = [
    "burger_king_jollof_youtube_comments.csv",
    "burger_king_jollof_analysed.csv",
]

print("=== COPYING BURGER KING FILES TO DESKTOP ===\n")

copied  = 0
skipped = 0

for filename in project_files:
    found = False
    for root, dirs, files in os.walk(os.path.expanduser("~")):
        for file in files:
            if file == filename:
                src  = os.path.join(root, file)
                dest = os.path.join(desktop, file)
                try:
                    shutil.copy2(src, dest)
                    print(f"✅ Copied: {filename}")
                    print(f"   📁 From: {src}")
                    print(f"   📁 To:   {dest}\n")
                    copied += 1
                    found = True
                    break
                except PermissionError:
                    print(f"⚠️ Permission Error: {filename}")
                    print(f"   Close the file if it is open in Excel then retry\n")
                    skipped += 1
                    found = True
                    break
                except Exception as e:
                    print(f"❌ Error copying {filename}: {e}\n")
                    skipped += 1
                    found = True
                    break
    if not found:
        print(f"❌ Not Found: {filename}")
        print(f"   Run the sentiment analysis code first to generate it\n")
        skipped += 1

print(f"Copied:  {copied} files ✅")
print(f"Skipped: {skipped} files ⚠️")
print(f"\nDesktop location: {desktop}")
```
